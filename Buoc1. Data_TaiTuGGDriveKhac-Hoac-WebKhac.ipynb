```python
from google.colab import drive
import os
import tarfile
```

```python
# 1. GẮN GOOGLE DRIVE
# Bước này cho phép Colab truy cập và ghi dữ liệu vào Drive của bạn
print("--- BƯỚC 1: GẮN DRIVE ---")
print("Nếu không thấy popup, kiểm tra VS Code Output panel (View > Output > Jupyter)")
drive.mount('/content/drive', force_remount=True)
```

```python
# 2. ĐỊNH NGHĨA ĐƯỜNG DẪN
# Thư mục đích trong Google Drive để lưu dữ liệu.
# Đảm bảo thư mục 'research' tồn tại trong My Drive.
BASE_DIR = '/content/drive/MyDrive/Colab Notebooks/DATA_Pancreas_Segmentation_Bao'

# Tên file TAR và URL nguồn
TAR_FILE_NAME = 'Task07_Pancreas.tar'
DOWNLOAD_URL = 'https://msd-for-monai.s3-us-west-2.amazonaws.com/Task07_Pancreas.tar'

# Đường dẫn file TAR tạm thời trong Colab storage (nhanh hơn Drive)
TEMP_DIR = '/content/temp_download'
TAR_FILE_PATH = os.path.join(TEMP_DIR, TAR_FILE_NAME)

# Thư mục giải nén cuối cùng trong Google Drive
EXTRACT_DIR = os.path.join(BASE_DIR, 'Task07_Pancreas_Raw')

# Tạo các thư mục cần thiết
os.makedirs(TEMP_DIR, exist_ok=True)  # Thư mục tạm trong Colab
os.makedirs(BASE_DIR, exist_ok=True)   # Thư mục đích trong Drive
os.makedirs(EXTRACT_DIR, exist_ok=True)

print(f"Thư mục tạm (Colab storage): {TEMP_DIR}")
print(f"Thư mục đích (Google Drive): {EXTRACT_DIR}")
```

```python
# 3. TẢI FILE TRỰC TIẾP TỪ URL VỀ COLAB STORAGE (TẠM THỜI)
print("\n--- BƯỚC 2: TẢI FILE TỪ URL VỀ COLAB STORAGE ---")
print(f"Bắt đầu tải file Task07_Pancreas.tar ({11.4} GB) về: {TAR_FILE_PATH}")
print("(Tải về Colab storage nhanh hơn nhiều so với tải trực tiếp vào Drive)")

# Sử dụng lệnh Wget để tải về thư mục tạm trong Colab
!wget -O "{TAR_FILE_PATH}" "{DOWNLOAD_URL}"

print("\n✅ Tải xuống hoàn tất!")
```

```python
# 4. GIẢI NÉN FILE .TAR TỪ COLAB STORAGE VÀO GOOGLE DRIVE
print("\n--- BƯỚC 3: GIẢI NÉN FILE VÀO GOOGLE DRIVE ---")
print(f"Nguồn (Colab storage): {TAR_FILE_PATH}")
print(f"Đích (Google Drive): {EXTRACT_DIR}")
print("Đang giải nén... (có thể mất vài phút)")

# Lệnh tar: -x (giải nén) -f (chỉ định file) -C (chỉ định thư mục đích)
!tar -xf "{TAR_FILE_PATH}" -C "{EXTRACT_DIR}"

print("\n✅ Giải nén hoàn tất!")
```

```python
# 5. DỌN DẸP - XÓA FILE .TAR TẠM TRONG COLAB STORAGE
print("\n--- BƯỚC 4: DỌN DẸP FILE .TAR TẠM ---")

# Xóa file tar tạm
if os.path.exists(TAR_FILE_PATH):
    os.remove(TAR_FILE_PATH)
    print(f"✅ Đã xóa file .tar tạm: {TAR_FILE_PATH}")

# Xóa thư mục tạm nếu rỗng
import shutil
if os.path.exists(TEMP_DIR):
    shutil.rmtree(TEMP_DIR)
    print(f"✅ Đã xóa thư mục tạm: {TEMP_DIR}")

print("\n✅ Dọn dẹp hoàn tất! File tar đã được xóa, dữ liệu đã lưu vào Drive.")
```
